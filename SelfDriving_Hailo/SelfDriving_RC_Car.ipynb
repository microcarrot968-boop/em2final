| **Right IR Sensor** | GPIO 10 | Right Line Detection |
| **Front IR Sensor** | GPIO 25 | Front Obstacle Detection |

## 🚗 **RC Car Controller Test**

**Use the modular approach (recommended):**

```bash
python3 rc_car_controller.py --test-servo
```

**Or run this cell for in-notebook motor testing:**
This tests basic motor and servo functionality.


````python
import RPi.GPIO as GPIO
import time
import numpy as np

# GPIO Pin Configuration
MOTOR_PWM_PIN = 18
MOTOR_DIR_PIN = 19
SERVO_PWM_PIN = 13

# Initialize GPIO
GPIO.setmode(GPIO.BCM)
GPIO.setup(MOTOR_PWM_PIN, GPIO.OUT)
GPIO.setup(MOTOR_DIR_PIN, GPIO.OUT)
GPIO.setup(SERVO_PWM_PIN, GPIO.OUT)

# Initialize PWM
motor_pwm = GPIO.PWM(MOTOR_PWM_PIN, 1000)  # 1kHz frequency
servo_pwm = GPIO.PWM(SERVO_PWM_PIN, 50)    # 50Hz for servo

motor_pwm.start(0)
servo_pwm.start(0)

def test_motor():
    """Test motor forward and reverse."""
    print("Testing motor...")
    
    # Forward
    GPIO.output(MOTOR_DIR_PIN, GPIO.HIGH)
    motor_pwm.ChangeDutyCycle(30)  # 30% speed
    time.sleep(2)
    
    # Stop
    motor_pwm.ChangeDutyCycle(0)
    time.sleep(1)
    
    # Reverse
    GPIO.output(MOTOR_DIR_PIN, GPIO.LOW)
    motor_pwm.ChangeDutyCycle(30)
    time.sleep(2)
    
    # Stop
    motor_pwm.ChangeDutyCycle(0)
    print("Motor test complete")

def test_servo():
    """Test servo steering."""
    print("Testing servo...")
    
    # Center position (1.5ms pulse = 7.5% duty cycle at 50Hz)
    servo_pwm.ChangeDutyCycle(7.5)
    time.sleep(1)
    
    # Left turn (1ms pulse = 5% duty cycle)
    servo_pwm.ChangeDutyCycle(5.0)
    time.sleep(1)
    
    # Right turn (2ms pulse = 10% duty cycle)
    servo_pwm.ChangeDutyCycle(10.0)
    time.sleep(1)
    
    # Center again
    servo_pwm.ChangeDutyCycle(7.5)
    print("Servo test complete")

try:
    test_motor()
    test_servo()
except Exception as e:
    print(f"Error during testing: {e}")
finally:
    # Cleanup
    motor_pwm.stop()
    servo_pwm.stop()
    GPIO.cleanup()
    print("GPIO cleanup complete")
````

## 📡 **Sensor Manager Test**

**Or run this cell for in-notebook sensor testing:**
This tests ultrasonic distance sensor and IR sensors.


````python
import RPi.GPIO as GPIO
import time
import numpy as np

# Sensor Pin Configuration
ULTRASONIC_TRIGGER = 23
ULTRASONIC_ECHO = 24
LEFT_IR = 9
RIGHT_IR = 10
FRONT_IR = 25

# Initialize GPIO
GPIO.setmode(GPIO.BCM)
GPIO.setup(ULTRASONIC_TRIGGER, GPIO.OUT)
GPIO.setup(ULTRASONIC_ECHO, GPIO.IN)
GPIO.setup(LEFT_IR, GPIO.IN)
GPIO.setup(RIGHT_IR, GPIO.IN)
GPIO.setup(FRONT_IR, GPIO.IN)

def read_ultrasonic():
    """Read distance from ultrasonic sensor."""
    # Send trigger pulse
    GPIO.output(ULTRASONIC_TRIGGER, GPIO.HIGH)
    time.sleep(0.00001)  # 10 microseconds
    GPIO.output(ULTRASONIC_TRIGGER, GPIO.LOW)
    
    # Wait for echo
    start_time = time.time()
    while GPIO.input(ULTRASONIC_ECHO) == 0:
        start_time = time.time()
    
    end_time = time.time()
    while GPIO.input(ULTRASONIC_ECHO) == 1:
        end_time = time.time()
    
    # Calculate distance (speed of sound = 34300 cm/s)
    duration = end_time - start_time
    distance = (duration * 34300) / 2
    
    return distance

def read_ir_sensors():
    """Read all IR sensors."""
    left = GPIO.input(LEFT_IR)
    right = GPIO.input(RIGHT_IR)
    front = GPIO.input(FRONT_IR)
    
    return {
        'left': left,
        'right': right,
        'front': front
    }

def test_ultrasonic():
    """Test ultrasonic sensor."""
    print("Testing ultrasonic sensor...")
    
    for i in range(5):
        distance = read_ultrasonic()
        print(f"Distance: {distance:.1f} cm")
        time.sleep(0.5)

def test_ir_sensors():
    """Test IR sensors."""
    print("Testing IR sensors...")
    print("Place objects near sensors to test detection")
    
    for i in range(10):
        ir_values = read_ir_sensors()
        print(f"IR - Left: {ir_values['left']}, Right: {ir_values['right']}, Front: {ir_values['front']}")
        time.sleep(0.5)

try:
    test_ultrasonic()
    test_ir_sensors()
except Exception as e:
    print(f"Error during sensor testing: {e}")
finally:
    GPIO.cleanup()
    print("Sensor testing complete")
````

## 🎯 **Object Detection Integration**

**Or run this cell for in-notebook camera and detection testing:**
This tests the camera and object detection integration.


````python
from picamera2 import Picamera2
import cv2
import numpy as np
import time

def test_camera():
    """Test camera functionality."""
    print("Testing camera...")
    
    picam2 = Picamera2()
    config = picam2.create_preview_configuration(
        main={'size': (640, 480), 'format': 'RGB888'}
    )
    picam2.configure(config)
    picam2.start()
    
    try:
        for i in range(5):
            frame = picam2.capture_array()
            print(f"Frame {i+1}: Shape {frame.shape}, Type {frame.dtype}")
            time.sleep(0.5)
    finally:
        picam2.stop()
    
    print("Camera test complete")

def test_object_detection():
    """Test object detection with camera."""
    print("Testing object detection...")
    
    # Import the object detector from Project 2
    try:
        from object_detector import ObjectDetector
        from model_manager import ModelManager
        
        # Setup model manager
        model_manager = ModelManager('models')
        if not model_manager.download_model():
            print("Failed to download model")
            return
        
        model_info = model_manager.get_model_info()
        
        # Initialize detector
        detector = ObjectDetector(
            model_path=model_info['model_path'],
            labels=model_info['labels'],
            csv_path='test_detection.csv'
        )
        
        # Test detection
        picam2 = Picamera2()
        config = picam2.create_preview_configuration(
            main={'size': (640, 480), 'format': 'RGB888'}
        )
        picam2.configure(config)
        picam2.start()
        
        try:
            for i in range(3):
                frame = picam2.capture_array()
                label, confidence, inference_time = detector.infer_frame(frame)
                print(f"Detection {i+1}: {label} (confidence: {confidence:.2f}, time: {inference_time:.1f}ms)")
                time.sleep(1)
        finally:
            picam2.stop()
        
        print("Object detection test complete")
        
    except ImportError:
        print("Object detector not available. Run Project 2 first.")
    except Exception as e:
        print(f"Error during object detection test: {e}")

try:
    test_camera()
    test_object_detection()
except Exception as e:
    print(f"Error during testing: {e}")
````

**Or run this cell for in-notebook autonomous driving test:**
This demonstrates the complete self-driving system.


````python
import time
import numpy as np
from datetime import datetime

def simulate_autonomous_driving():
    """Simulate autonomous driving decisions."""
    print("=== Self-Driving System Simulation ===")
    print("This is a simulation of the autonomous driving system")
    print("In actual implementation, this would control real motors and sensors")
    
    # Simulate sensor readings
    def simulate_sensors():
        return {
            'distance': np.random.uniform(10, 100),  # cm
            'left_ir': np.random.choice([0, 1]),
            'right_ir': np.random.choice([0, 1]),
            'front_ir': np.random.choice([0, 1]),
            'detected_object': np.random.choice(['person', 'car', 'bicycle', 'none']),
            'object_confidence': np.random.uniform(0.3, 0.95)
        }
    
    # Simulate driving decisions
    def make_driving_decision(sensors):
        speed = 0.5  # Default speed
        steering = 0.0  # Center steering
        action = "forward"
        
        # Obstacle avoidance logic
        if sensors['distance'] < 20:
            speed = 0.0
            action = "stop"
        elif sensors['distance'] < 40:
            speed = 0.2
            action = "slow"
        
        # Person detection
        if sensors['detected_object'] == 'person' and sensors['object_confidence'] > 0.7:
            speed = 0.0
            action = "stop_for_person"
        
        # Line following logic
        if sensors['left_ir'] == 1 and sensors['right_ir'] == 0:
            steering = -0.3  # Turn right
            action = "turn_right"
        elif sensors['left_ir'] == 0 and sensors['right_ir'] == 1:
            steering = 0.3   # Turn left
            action = "turn_left"
        
        return {
            'speed': speed,
            'steering': steering,
            'action': action
        }
    
    # Run simulation
    print("\nStarting autonomous driving simulation...")
    print("Time\tDistance\tIR_L\tIR_R\tObject\tConf\tSpeed\tSteering\tAction")
    print("-" * 80)
    
    for i in range(20):
        sensors = simulate_sensors()
        decision = make_driving_decision(sensors)
        
        print(f"{i+1:2d}\t{sensors['distance']:6.1f}\t{sensors['left_ir']}\t{sensors['right_ir']}\t"
              f"{sensors['detected_object']:8s}\t{sensors['object_confidence']:.2f}\t"
              f"{decision['speed']:.1f}\t{decision['steering']:7.1f}\t{decision['action']}")
        
        time.sleep(0.5)
    
    print("\nSimulation complete!")
    print("\nIn the actual implementation:")
    print("- Speed values would control motor PWM duty cycle")
    print("- Steering values would control servo angle")
    print("- Sensor readings would come from real hardware")
    print("- Object detection would use TensorFlow Lite model")
    print("- All decisions would be logged to CSV files")

try:
    simulate_autonomous_driving()
except Exception as e:
    print(f"Error during simulation: {e}")
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

def analyze_driving_performance():
    """Analyze driving performance data."""
    print("=== Driving Performance Analysis ===")
    
    # Create sample driving data
    np.random.seed(42)
    n_samples = 100
    
    data = {
        'timestamp': [datetime.now() - timedelta(seconds=i) for i in range(n_samples)],
        'speed': np.random.uniform(0, 0.8, n_samples),
        'steering_angle': np.random.uniform(-0.5, 0.5, n_samples),
        'distance_front': np.random.uniform(5, 150, n_samples),
        'detected_object': np.random.choice(['person', 'car', 'bicycle', 'none'], n_samples),
        'object_confidence': np.random.uniform(0.1, 0.95, n_samples),
        'action_taken': np.random.choice(['forward', 'turn_left', 'turn_right', 'stop', 'slow'], n_samples),
        'safety_event': np.random.choice([0, 1], n_samples, p=[0.9, 0.1])
    }
    
    df = pd.DataFrame(data)
    
    print(f"\nDataset Overview:")
    print(f"Total samples: {len(df)}")
    print(f"Time range: {df['timestamp'].min()} to {df['timestamp'].max()}")
    print(f"Average speed: {df['speed'].mean():.2f}")
    print(f"Average distance: {df['distance_front'].mean():.1f} cm")
    
    # Performance metrics
    print(f"\nPerformance Metrics:")
    print(f"Safety events: {df['safety_event'].sum()} ({df['safety_event'].mean()*100:.1f}%)")
    print(f"Average object confidence: {df['object_confidence'].mean():.2f}")
    print(f"Speed variance: {df['speed'].var():.3f}")
    
    # Action distribution
    print(f"\nAction Distribution:")
    action_counts = df['action_taken'].value_counts()
    for action, count in action_counts.items():
        print(f"{action}: {count} ({count/len(df)*100:.1f}%)")
    
    # Create visualizations
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    
    # Speed over time
    axes[0, 0].plot(df.index, df['speed'])
    axes[0, 0].set_title('Speed Over Time')
    axes[0, 0].set_xlabel('Sample')
    axes[0, 0].set_ylabel('Speed')
    
    # Distance distribution
    axes[0, 1].hist(df['distance_front'], bins=20, alpha=0.7)
    axes[0, 1].set_title('Distance Distribution')
    axes[0, 1].set_xlabel('Distance (cm)')
    axes[0, 1].set_ylabel('Frequency')
    
    # Steering angle
    axes[1, 0].plot(df.index, df['steering_angle'])
    axes[1, 0].set_title('Steering Angle Over Time')
    axes[1, 0].set_xlabel('Sample')
    axes[1, 0].set_ylabel('Steering Angle')
    
    # Object detection confidence
    axes[1, 1].hist(df['object_confidence'], bins=20, alpha=0.7)
    axes[1, 1].set_title('Object Detection Confidence')
    axes[1, 1].set_xlabel('Confidence')
    axes[1, 1].set_ylabel('Frequency')
    
    plt.tight_layout()
    plt.show()
    
    # Correlation analysis
    print(f"\nCorrelation Analysis:")
    numeric_cols = ['speed', 'steering_angle', 'distance_front', 'object_confidence']
    corr_matrix = df[numeric_cols].corr()
    
    plt.figure(figsize=(8, 6))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)
    plt.title('Correlation Matrix')
    plt.show()
    
    print(f"\nKey Correlations:")
    print(f"Speed vs Distance: {corr_matrix.loc['speed', 'distance_front']:.3f}")
    print(f"Steering vs Distance: {corr_matrix.loc['steering_angle', 'distance_front']:.3f}")
    print(f"Speed vs Confidence: {corr_matrix.loc['speed', 'object_confidence']:.3f}")
    
    return df

try:
    driving_data = analyze_driving_performance()
    print("\nAnalysis complete!")
except Exception as e:
    print(f"Error during analysis: {e}")
````
